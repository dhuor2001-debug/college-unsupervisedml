## 1. Association Rule Mining using Apriori Algorithm

### Problem 1
Find the frequent itemsets and association rules from the given transactional dataset using **minimum support = 60%** and **minimum confidence = 70%**.

```python
%pip install mlxtend
```

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
from mlxtend.preprocessing import TransactionEncoder

# 1. Define sample transactional data
dataset = [
    ["Milk", "Bread", "Butter"],
    ["Bread", "Butter", "Eggs"],
    ["Milk", "Bread", "Eggs"],
    ["Milk", "Bread", "Butter", "Eggs"],
    ["Milk", "Butter"],
]

# 2. One-hot encode the data into a boolean DataFrame
te = TransactionEncoder()
te_ary = te.fit(dataset).transform(dataset)
df = pd.DataFrame(te_ary, columns=te.columns_)
print(df)

# 3. Find frequent itemsets using the Apriori algorithm
frequent_itemsets = apriori(df, min_support=0.6, use_colnames=True)
print("\nFrequent Itemsets:")
print(frequent_itemsets)

# 4. Generate association rules
result_rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.7
)
print("\nAssociation Rules:")
print(result_rules)
```

### Problem 2
Apply the Apriori algorithm to the following transactional dataset using **minimum support = 50%** and **minimum confidence = 70%**.

## 2. Hierarchical Clustering (Bottom-Up Approach)

```python
from sklearn.cluster import AgglomerativeClustering
import matplotlib.pyplot as plt

x = [4, 5, 10, 4, 3, 11, 14, 6, 10, 12]
y = [21, 19, 24, 17, 16, 25, 24, 22, 21, 21]
data = list(zip(x, y))

# Initialize hierarchical clustering with 2 clusters and Ward linkage
hierarchical_cluster = AgglomerativeClustering(
    n_clusters=2,
    linkage='ward'
)

labels = hierarchical_cluster.fit_predict(data)

# Plot the original data points color-coded by cluster
plt.scatter(x, y, c=labels, label='Data Points')
plt.title('Hierarchical Clustering')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend(loc='lower right')
plt.show()
```

**Note:** `linkage='ward'` is a hierarchical agglomerative clustering method that merges pairs of clusters to minimize the increase in total within-cluster variance.

### Plot Dendrogram for Hierarchical Clustering

```python
from scipy.cluster.hierarchy import dendrogram, linkage
import matplotlib.pyplot as plt

x = [4, 5, 10, 4, 3, 11, 14, 6, 10, 12]
y = [21, 19, 24, 17, 16, 25, 24, 22, 21, 21]
data = list(zip(x, y))

# Compute linkage using Euclidean distance and Ward linkage
linkage_data = linkage(data, method='ward', metric='euclidean')

# Plot dendrogram
dendrogram(linkage_data)
plt.title('Dendrogram for Hierarchical Clustering')
plt.ylabel('Height')
plt.show()
```

**Notes:**
- `dendrogram()` plots the hierarchical clustering represented by a linkage matrix as a tree diagram.
- `plt.show()` visualizes the hierarchy from individual points at the bottom to a single cluster at the top.
- A dendrogram is a tree-like chart showing the step-by-step merging of data points in hierarchical clustering.

## 3. K-Means Clustering

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

x = [4, 5, 10, 4, 3, 11, 14, 6, 10, 12]
y = [21, 19, 24, 17, 16, 25, 24, 22, 21, 21]
data = list(zip(x, y))

kmeans = KMeans(n_clusters=2)
kmeans.fit(data)

# Plot the original data points color-coded by cluster
plt.scatter(x, y, c=kmeans.labels_, label='Data Points')
plt.title('K-Means Clustering')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend(loc='lower right')
plt.show()
```

**Notes:**
- `list(zip(x, y))` pairs the coordinates into 2D points.
- `KMeans(n_clusters=2)` initializes K-Means to find 2 groups.
- `kmeans.fit(data)` computes the cluster centers and assigns points to clusters.
- `kmeans.labels_` contains the cluster label (0 or 1) for each point.

### K-Means Clustering with Centroids

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

x = [4, 5, 10, 4, 3, 11, 14, 6, 10, 12]
y = [21, 19, 24, 17, 16, 25, 24, 22, 21, 21]
data = list(zip(x, y))

kmeans = KMeans(n_clusters=2)
kmeans.fit(data)

plt.scatter(x, y, c=kmeans.labels_, label='Data Points')

# Get the centroid coordinates
centroids = kmeans.cluster_centers_

# Plot the centroids
plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    c='red',
    marker='*',
    s=200,
    label='Centroids'
)

plt.title('K-Means Clustering with Centroids')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend(loc='lower right')
plt.show()

print("Centroids:")
print(centroids)
```

**Note:** `kmeans.cluster_centers_` extracts the calculated center coordinates. The supplied lab notes give the example centroid coordinates as approximately `[[4.4, 19.0], [11.4, 23.0]]`.

## 4. Dimensionality Reduction by PCA

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Load Iris dataset as a pandas DataFrame
iris = load_iris(as_frame=True)
df = iris.frame

# View the first few rows
print(df.head())
```

```python
# 1. Load a sample dataset (Iris Dataset: 4 features, 3 classes)
iris = load_iris()
X = iris.data
y = iris.target
feature_names = iris.feature_names
target_names = iris.target_names

print("Original shape:", X.shape)  # (150, 4)

# 2. Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Initialize and fit PCA
# Reduce from 4 dimensions down to 2 principal components
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Reduced shape:", X_pca.shape)  # (150, 2)

# 4. Visualize the 2D projection
plt.figure(figsize=(8, 6))
colors = ["navy", "turquoise", "darkorange"]

for color, i, target_name in zip(colors, [0, 1, 2], target_names):
    plt.scatter(
        X_pca[y == i, 0],
        X_pca[y == i, 1],
        color=color,
        label=target_name,
    )

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA of IRIS Dataset")
plt.legend(loc="best")
plt.grid(True)
plt.show()
```
